```python
# 02_flight_model_training.ipynb

from pathlib import Path
import mlflow

# Make sure project root is on sys.path
import sys, os
PROJECT_ROOT = Path.cwd().parent  # notebooks/ → project root
if str(PROJECT_ROOT) not in sys.path:
    sys.path.append(str(PROJECT_ROOT))

print("PROJECT_ROOT:", PROJECT_ROOT)

from src.data_handler import load_csv
from src.data_wrangling import wrangle_flights
from src.data_preprocessing import preprocess_flights, data_spliting_Xy, data_spliting_X
from src.training import train_flight_price_model
from src.prediction import predict_flight_price
```

```python
RAW_DIR = PROJECT_ROOT / "data" / "raw"

flights = load_csv(RAW_DIR / "flights.csv")
print("Flights wrangled shape:", flights.shape)
display(flights.head())
```

```python
# Data Wrangling
flights_wrangled = wrangle_flights(flights)
print("Flights_wrangled shape:", flights_wrangled.shape)
display(flights_wrangled.head())
```

```python
# Full preprocessing + feature engineering
flights_prep = preprocess_flights(flights_wrangled)
print("Flights_prep shape:", flights_prep.shape)
display(flights_prep.head())

# Build feature matrix and target
X_flight, y_flight = data_spliting_Xy(flights_prep, target_col = "price", drop_cols = ["travelcode", "usercode", "flighttype", "time", "distance", "date"])
                                                                                    
print("X_flight shape:", X_flight.shape)
print("y_flight shape:", y_flight.shape)
display(X_flight.head())
display(y_flight.head())
```

```python
mlflow.set_tracking_uri("http://127.0.0.1:5000")  # adjust if needed

results = train_flight_price_model(experiment_name = "voyage_flight_price")

best_model_name = results["best_model"]
print("\nBest model:", best_model_name)
print("Best test RMSE:", results["results"][best_model_name]["test_rmse"])
print("Best test R² :", results["results"][best_model_name]["test_r2"])
```

```python
for name, res in results["results"].items():
    print(f"\nModel: {name}")
    print(f"  Train RMSE: {res['train_rmse']:.2f}")
    print(f"  Test  RMSE: {res['test_rmse']:.2f}")
    print(f"  Train R² : {res['train_r2']:.3f}")
    print(f"  Test  R² : {res['test_r2']:.3f}")
```

```python
# Choose the best model file
model_path = PROJECT_ROOT / "models" / f"{best_model_name}_flight.pkl"
print("Using model:", model_path)

sample_input = {
    "from": flights["from"].iloc[0],
    "to": flights["to"].iloc[0],
    "flighttype": flights["flightType"].iloc[0],
    "agency": flights["agency"].iloc[0],
    "distance": float(flights["distance"].iloc[0]),
    "time": float(flights["time"].iloc[0]),
    "date": str(flights["date"].iloc[0]) if "date" in flights.columns else "2026-01-15",
}

prediction_result = predict_flight_price(
    model_path=model_path,
    input_data=sample_input,
)

prediction_result
```
